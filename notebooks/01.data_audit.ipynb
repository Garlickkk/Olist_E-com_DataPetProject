```python
import matplotlib.pyplot as plt

orders["purchase_month"] = orders["order_purchase_timestamp"].dt.to_period("M")
monthly_all = orders.groupby("purchase_month").size()
monthly_delivered = orders[orders["order_status"] == "delivered"].groupby("purchase_month").size()

monthly_df = pd.DataFrame({
    "Все заказы": monthly_all,
    "Доставленные": monthly_delivered
}).fillna(0)

ax = monthly_df.plot(kind="bar", figsize=(13, 5), width=0.8)
plt.title("Объём заказов по месяцам")
plt.xlabel("Месяц покупки")
plt.ylabel("Количество заказов")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

monthly_df
```

```python
delivered = orders[
    (orders["order_status"] == "delivered") & 
    (orders["order_delivered_customer_date"].notna())
].copy()

delivered["delivery_days"] = (
    delivered["order_delivered_customer_date"] - delivered["order_purchase_timestamp"]
).dt.total_seconds() / 86400

delivered["delay_days"] = (
    delivered["order_delivered_customer_date"].dt.normalize() - 
    delivered["order_estimated_delivery_date"].dt.normalize()
).dt.days

print("Статистика по срокам доставки (delivery_days) и отклонению от прогноза (delay_days):")
display(delivered[["delivery_days", "delay_days"]].describe(percentiles=[0.05, 0.25, 0.5, 0.75, 0.95, 0.99]))

late_share = (delivered["delay_days"] > 0).mean() * 100
print(f"Опоздавших заказов: {(delivered['delay_days'] > 0).sum()} ({late_share:.2f}%)")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(delivered["delivery_days"], bins=50, range=(0, 60), edgecolor="black")
axes[0].axvline(delivered["delivery_days"].median(), color="red", linestyle="--", label=f"Медиана: {delivered['delivery_days'].median():.1f} дн.")
axes[0].set_title("Распределение фактического срока доставки (0–60 дней)")
axes[0].set_xlabel("Дней от покупки до получения")
axes[0].set_ylabel("Количество заказов")
axes[0].legend()
axes[0].grid(axis="y", linestyle="--", alpha=0.5)


axes[1].hist(delivered["delay_days"], bins=50, range=(-40, 40), edgecolor="black")
axes[1].axvline(0, color="red", linestyle="--", label="Граница опоздания (0 дней)")
axes[1].set_title("Отклонение от обещанной даты (+-40 дней)")
axes[1].set_xlabel("Дней (< 0 — раньше срока, > 0 — опоздание)")
axes[1].set_ylabel("Количество заказов")
axes[1].legend()
axes[1].grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
